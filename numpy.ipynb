```python
import time
import numpy as np
```

```python
start = time.time()
numpy_arr= np.arange(1000000)
numpy_arr = numpy_arr**2
end = time.time()
numpy_time = end-start
print(f"Time taken: {numpy_time:.6f} seconds")
```

```python
import time
start = time.time()
list1= list(range(1000000))
for i in list1:
    i*=i
end = time.time()
list_time = end-start
print(f"Time taken: {list_time:.6f} seconds")
```

```python
speedup = list_time/numpy_time
print(speedup)
```
